```python
ufc_fighters = ufc_fighters.drop(['Nickname'], axis=1)
```

```python
# Add age in Days Column
ufc_fighters['DOB'] = pd.to_datetime(ufc_fighters['DOB'], errors='coerce')
ufc_fighters['age_d'] = (pd.Timestamp('today') - ufc_fighters['DOB']).dt.days
ufc_fighters
```

```python
ufc_fighters.dtypes
```

```python
categorical_columns = ['Stance']
ufc_fighters = pd.get_dummies(ufc_fighters, columns=categorical_columns)
ufc_fighters.dtypes
```

```python
stats_columns = ['Streak', 'Height', 'Weight', 'Reach', 'Sig_Str_M', 'Sig_Str_Acc',
                'Sig_Str_Abs_M', 'Str_Def', 'TD_Avg', 'TD_Acc', 'TD_Def', 'Sub_Avg', 'Wins', 'Losses',
                'age_d', 'Stance_Open Stance','Stance_Orthodox', 'Stance_Sideways', 'Stance_Southpaw',
                 'Stance_Switch']

```

```python
from sklearn.metrics import pairwise_distances
from sklearn.preprocessing import StandardScaler

ufc_fighters = ufc_fighters.dropna()

scaler = StandardScaler()

X_scaled = scaler.fit_transform(ufc_fighters[stats_columns])


```

```python
ufc_fighters.isna().sum()
```

```python
# Compute the pairwise distance matrix
distance_matrix = pairwise_distances(X_scaled, metric='euclidean')

# Convert the distance matrix to a DataFrame for better readability
distance_df = pd.DataFrame(distance_matrix, index=ufc_fighters['Name'], columns=ufc_fighters['Name'])

# Display the pairwise distance matrix
distance_df
```

```python
np.fill_diagonal(distance_matrix, np.inf)
```

```python
distance_df.to_csv('fighter_alike_distance.csv')
```

```python

def find_alike(name):
    """

    :param Name: Name of fighter
    :return: prints a list of alike fighters if the fighter exist
    """

    fighter = distance_df[distance_df.index.str.contains(name, case=False)]
    if not fighter.empty:
        fighter_name = fighter.index[0]
        fighter_distances = distance_df.loc[fighter_name]
        closest_fighters = fighter_distances.sort_values().iloc[1:21]
        print(f"Closest fighters to {fighter_name}:")
        print(closest_fighters)
    else:
        print(f"No fighter found matching the name '{name}'.")

```

```python
merab_alike = find_alike(distance_df, 'Merab')
```

```python
omalley_alike = find_alike('O\'Malley')
```

```python
def compare_fighters(df, f1, f2):
    """
    Compare two fighters and return the distance (similarity) between them.

    :param df: DataFrame of fighter pairwise distances.
    :param f1: Name of the first fighter.
    :param f2: Name of the second fighter.
    :return: Distance between the two fighters or an error message.
    """
    # Ensure the index is of string type
    df.index = df.index.astype(str)

    # Ensure f1 and f2 are in the dataset
    if f1 in df.index and f2 in df.columns:
        distance = df.loc[f1, f2]
        return distance
    else:
        return 'One or both fighters were not found in the dataset.'
```
