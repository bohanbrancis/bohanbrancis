```python
from imdb import Cinemagoer
ia = Cinemagoer()



# Search for a movie by title
movies = ia.search_movie('Avengers Endgame')

# Assuming the first search result is the desired movie
the_matrix = movies[0]

# Get detailed information of the movie
ia.update(the_matrix)

# Access the box office information, if available
if 'box office' in the_matrix.infoset2keys:
    box_office_info = the_matrix.get('box office')
    for key, value in box_office_info.items():
        print(f"{key}: {value}")
else:
    print("Box office information is not available.")
```
